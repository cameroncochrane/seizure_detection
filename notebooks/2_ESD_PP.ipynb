<h3>Epileptic Seizure Detection: Notebook 2: Pre-processing and Feature Selection Pipeline Implementation

```python
import pickle as pkl
from pathlib import Path

PROJ_ROOT = Path().resolve().parents[0]
DATA_DIR = PROJ_ROOT / "data"
RAW_DATA_DIR = DATA_DIR / "raw"
RAW_DATA_PATH = RAW_DATA_DIR / "BEED_Data.csv"
INTERIM_DATA_DIR = DATA_DIR / "interim"
INTERIM_DATA_PATH = INTERIM_DATA_DIR / "data_outliers_removed.pkl"

import pandas as pd
import numpy as np

with open(INTERIM_DATA_PATH, 'rb') as f:
    data = pkl.load(f)
```

```python
data.head(10)
```

Let's train-test split the data now ready for the creation of the pipeline

```python
from sklearn.model_selection import train_test_split

# Separate features and target
X = data.drop('y', axis=1)
y = data['y']

# Stratified train-test split
X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.2, random_state=13, stratify=y
)
```

<h4>Pipeline Creation

Steps to carry out:
1. Feature selection (features to remove already identified).
2. Scaling of features.
3. Necessary encoding of target.
4. Sampling.
5. Modelling.

- All carried out using 5-10 fold cross validation
